```python
import numpy as np

x = np.array([-1.650680, -0.524648,
               0.524648,  1.650680])

w = np.array([0.081313, 0.804914,
              0.804914, 0.081313])

def f(x):
    return x**2 + 2*x + 1

I = np.sum(w * f(x))

print("Gauss-Hermite Quadrature")
print("Number of points = 4")
print("Roots =", x)
print("Weights =", w)
print("Integration =", I)
```
