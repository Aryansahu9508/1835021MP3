```python
import numpy as np

f=np.array([1,2,3,4])
n=len(f)
F_k=np.zeros(n,dtype=complex)
F_I=np.zeros(n,dtype=complex)

for k in range(n):
    for j in range(n):
        F_k[k]+=f[j]*np.exp(-1j*2*np.pi*k*j/n)
for i in range(n):
    for k in range(n):
        F_I[i]+=F_k[k]*np.exp(1j*2*np.pi*k*i/n )

print(F_k)
print(F_I)
```
