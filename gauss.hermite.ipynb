```python
import numpy as np

x = np.array([-1.650680,
              -0.524648,
               0.524648,
               1.650680])
w = np.array([0.081313,
              0.804914,
              0.804914,
              0.081313])

def f(x):
    return x**2

I = 0

for i in range(4):
    I = I + w[i] * f(x[i])

# Display result
print("Gauss-Hermite Integration")
print("Number of points = 4")
print("Value of integration =", I)
```
