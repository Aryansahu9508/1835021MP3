```python

import numpy as np

# Boundary conditions
a = 0
b = np.pi / 2
ya = 0
yb = 1

# Step size
h = 0.01
N = int((b-a)/h)

# RK4 method
def solve_shooting(s):

    x = np.linspace(a, b, N+1)

    y = np.zeros(N+1)
    z = np.zeros(N+1)

    y[0] = ya
    z[0] = s

    for i in range(N):

        k1 = h*z[i]
        l1 = -h*y[i]

        k2 = h*(z[i]+l1/2)
        l2 = -h*(y[i]+k1/2)

        k3 = h*(z[i]+l2/2)
        l3 = -h*(y[i]+k2/2)

        k4 = h*(z[i]+l3)
        l4 = -h*(y[i]+k3)

        y[i+1] = y[i] + (k1+2*k2+2*k3+k4)/6
        z[i+1] = z[i] + (l1+2*l2+2*l3+l4)/6

    return x, y

# Secant method
s0 = 0.5
s1 = 1.5

for i in range(20):

    x, y0 = solve_shooting(s0)
    x, y1 = solve_shooting(s1)

    f0 = y0[-1] - yb
    f1 = y1[-1] - yb

    s2 = s1 - f1*(s1-s0)/(f1-f0)

    if abs(s2-s1) < 1e-10:
        break

    s0 = s1
    s1 = s2

# Final solution
x, y = solve_shooting(s2)

print("Initial slope =", s2)
print("Numerical boundary value =", y[-1])
print("Exact boundary value =", 1.0)

print("\nNumerical Solution:")
for i in range(0, len(x), 10):
    print("x =", round(x[i], 3),
          " y =", round(y[i], 6))
```
