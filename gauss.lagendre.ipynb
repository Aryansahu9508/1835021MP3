```python
import numpy as np

def f(x):
    return x**2

a = 0
b = 2
n = 2

x, w = np.polynomial.legendre.leggauss(n)

s = 0

for i in range(n):
    t = ((b-a)*x[i] + (b+a))/2
    s = s + w[i]*f(t)

I = (b-a)/2 * s

print("Integral =", I)
```
